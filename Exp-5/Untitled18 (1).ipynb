```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

df = pd.read_csv("EDA_Data_Visualization.csv")

df.head()
```

```python
x = np.linspace(0, 10, 100)
y = np.exp(-0.25 * x) * np.cos(x)

plt.plot(x, y, color='green', label='Damped Wave')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Line Plot')
plt.legend()
plt.show()
```

```python
categories = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
values = [42, 28, 35, 18, 12]

plt.bar(categories, values, color='teal')
plt.xlabel('Month')
plt.ylabel('Sales (in thousands)')
plt.title('Monthly Sales')
plt.show()
```

```python
data = np.random.normal(loc=60, scale=8, size=800)

plt.hist(data, bins=20, color='skyblue', edgecolor='black')
plt.xlabel('Marks')
plt.ylabel('Number of Students')
plt.title('Distribution of Marks')
plt.show()
```
