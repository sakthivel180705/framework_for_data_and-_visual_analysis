```python
import pandas as pd

print("========== IMPORT CSV FILE ==========")

df_csv = pd.read_csv("Student.csv")
print(df_csv.head())

print("\n========== IMPORT EXCEL FILE ==========")

df_excel = pd.read_excel("Student.xlsx")
print(df_excel.head())

print("\n========== IMPORT TABLE FROM WEBSITE ==========")

url = "https://en.wikipedia.org/wiki/List_of_Indian_states_and_union_territories_by_GDP"

tables = pd.read_html(url)

df_web = tables[0]

print(df_web.head())

print("\n========== EXPORT TO EXCEL ==========")

df_excel.to_excel("Student_Output.xlsx", index=False)

print("Student_Output.xlsx created successfully.")
```

```python
import sqlite3
import pandas as pd


conn = sqlite3.connect("Student.db")


data = {
    "RollNo": [101, 102, 103, 104, 105],
    "Name": ["Rahul", "Anita", "John", "Priya", "Arun"],
    "Department": ["CSE", "ECE", "IT", "AI", "CSE"],
    "Marks": [87, 92, 78, 95, 81]
}

df = pd.DataFrame(data)


df.to_sql("Student", conn, if_exists="replace", index=False)

conn.close()

print("Student.db created successfully.")
```

```python
import pandas as pd
import sqlite3

print("========== CSV FILE ==========")

df_csv = pd.read_csv("Student.csv")
print(df_csv)

print("\n========== EXCEL FILE ==========")

df_excel = pd.read_excel("Student.xlsx")
print(df_excel)

print("\n========== SQL FILE ==========")

conn = sqlite3.connect("Student.db")

df_sql = pd.read_sql("SELECT * FROM Student", conn)

print(df_sql)

conn.close()
```

```python
df_sql.to_excel("Student_Output.xlsx", index=False)

print("Student_Output.xlsx created successfully.")
```
